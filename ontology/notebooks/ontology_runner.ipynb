```python
# ------------------------------------------
# Ontology Runner
# ------------------------------------------

from ontology.ontology_loader import load_ontology

from ontology.entities.entity_manager import (
    get_entities,
    get_entity_by_name
)

from ontology.entities.entity_data_manager import (
    get_entity_dataframe
)

from ontology.relationships.relationship_manager import (
    get_relationships,
    get_relationship_by_name
)

from ontology.relationships.relationship_data_manager import (
    join_entities
)

from ontology.metadata.kpi_manager import (
    get_kpis
)

from ontology.metadata.business_term_manager import (
    get_business_terms
)

from ontology.knowledge_graph_manager import (
    build_graph
)

# ------------------------------------------
# Load Ontology
# ------------------------------------------

ontology = load_ontology()

print("=" * 50)
print("Ontology Name :", ontology["ontology_name"])
print("Version       :", ontology["version"])
print("=" * 50)

# ------------------------------------------
# Entities
# ------------------------------------------

print("\nEntities")

entities = get_entities(ontology)

for entity in entities:
    print(f"- {entity['name']}")

# ------------------------------------------
# Relationships
# ------------------------------------------

print("\nRelationships")

relationships = get_relationships(ontology)

for relationship in relationships:
    print(f"- {relationship['name']}")

# ------------------------------------------
# Business Terms
# ------------------------------------------

print("\nBusiness Terms")

business_terms = get_business_terms(ontology)

for term in business_terms:
    print(f"- {term['name']}")

# ------------------------------------------
# KPIs
# ------------------------------------------

print("\nKPIs")

kpis = get_kpis(ontology)

for kpi in kpis:
    print(f"- {kpi['name']}")

# ------------------------------------------
# Customer Data
# ------------------------------------------

print("\nFetching Customer Data...")

customer_entity = get_entity_by_name(
    ontology,
    "Customer"
)

customer_df = get_entity_dataframe(
    spark,
    customer_entity
)

display(customer_df)

# ------------------------------------------
# Customer Metrics Data
# ------------------------------------------

print("\nFetching Customer Metrics Data...")

metrics_entity = get_entity_by_name(
    ontology,
    "CustomerMetrics"
)

metrics_df = get_entity_dataframe(
    spark,
    metrics_entity
)

display(metrics_df)

# ------------------------------------------
# Join Using Ontology Relationship
# ------------------------------------------

print("\nJoining Customer and CustomerMetrics...")

relationship = get_relationship_by_name(
    ontology,
    "customer_has_metrics"
)

joined_df = join_entities(
    customer_df,
    metrics_df,
    relationship
)

display(joined_df)

# ------------------------------------------
# Build Knowledge Graph
# ------------------------------------------

print("\nKnowledge Graph")

nodes, edges = build_graph(ontology)

print("Nodes:")
print(nodes)

print("\nEdges:")
print(edges)
```
